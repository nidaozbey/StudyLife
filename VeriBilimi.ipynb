##Kayıp Veri (Missing Veri)
Amaç:

 1. Eksik veri tespiti ve farklı eksik veri çözüm yöntemlerini uygulama
 2. Eğitimde kullanılacak güncel veri setini kaydetme

 Adımlar:

  1. Veri setini yükleme
  2. eksik verileri tespit etme
  3. eksik veri oranlarını hesaplama
  4. eksik veri içeren satırları incele
  5. eksik verileri sil
  6. sayısal eksik verileri ortalama ile doldur
  7. sayısal eksik verileri medyan ile doldur
  8. sayısal eksik verileri mod ile doldur
  9. kategorik eksik verileri mod ile doldur
  10. kategorik eksik verileri sabit bir deger ile doldur
  11. eğitimde kullanılacak olan güncel veri setini oluştur

```python
import pandas as pd
```

```python
import pandas as pd
df = pd.read_csv("e_ticaret_veri_seti (1).csv")
display(df.head())
```

```python
#1.veri setini yükle
df=pd.read_csv("e_ticaret_veri_seti (1).csv")
```

```python
#2. eksik verileri tespit etme
df.isnull().sum()
```

```python
#eksik veri oranlarını hesapla
eksik_sayilari=df.isnull().sum()
eksik_oranlari=100 * (eksik_sayilari / len(df))
eksik_oranlari
```

```python
# 4. Eksik veri içeren satırları incele
eksik_satırlar = df[df.isnull().any(axis=1)]
eksik_satırlar.head(10) #ilk 10 satır
```

```python
#eksik verileri silme
df_silinen=df.dropna() #dropna ile eksik değer içeren satırları veri setinden çıkardık
```

```python
df_silinen.shape
```

```python
# Sayısal eksik verilerin ortalama ile doldurulması
df_ortalama = df.copy()

sayisal_sutunlar = df_ortalama.select_dtypes(
    include=["float64", "int64"]
).columns

for sutun in sayisal_sutunlar:
    df_ortalama[sutun] = df_ortalama[sutun].fillna(
        df_ortalama[sutun].mean()
    )
```

```python
df_ortalama[sayisal_sutunlar].isnull().sum()
```

```python
# 7. sayısal eksik verileri medyanla doldurma
df_medyan=df.copy()
for sutun in sayisal_sutunlar:
  df_medyan[sutun]=df_medyan[sutun].fillna(df_medyan[sutun].median())
  df_medyan[sayisal_sutunlar].isnull().sum()
```

```python
# 8. sayısal eksik verileri modla doldurma
df_mod=df.copy()
for sutun in sayisal_sutunlar:
  df_mod[sutun]=df_mod[sutun].fillna(df_mod[sutun].mode()[0])
  df_mod[sayisal_sutunlar].isnull().sum()
```

```python
#9. kategorik eksik verileri mod ile doldurma
df_mod_kategorik=df.copy()
kategorik_sutunlar=df_mod_kategorik.select_dtypes(include=["object"]).columns
kategorik_sutunlar
for sutun in kategorik_sutunlar:
  mod_degeri=df_mod_kategorik[sutun].mode()[0]
  df_mod_kategorik[sutun]=df_mod_kategorik[sutun].fillna(mod_degeri)
```

```python
df_mod_kategorik[kategorik_sutunlar].isnull().sum()
```

```python
#10. kategorik eksik verileri sabit bir degerle dolduralım
df_sabit=df.copy()
for sutun in kategorik_sutunlar:
  sabit_deger="belirtilmemiş"
  df_sabit[sutun]=df_sabit[sutun].fillna(sabit_deger)
```

```python
df_sabit[kategorik_sutunlar].isnull().sum()
```

```python
#11. eğitimde kullanacagımız güncel veri seti
# - sayısal verileri medyanla doldur
# + kategorik verileri mod ile doldur
df_guncel=df.copy()
for sutun in sayisal_sutunlar:
  df_guncel[sutun]=df_guncel[sutun].fillna(df_guncel[sutun].median)
```

```python
for sutun in kategorik_sutunlar:
   mod_degeri=df_guncel[sutun].mode()[0]
   df_guncel[sutun]=df_guncel[sutun].fillna(mod_degeri)
```

```python
df_guncel.isnull().sum()
```

```python
#12. düzenlenmiş veri setini kaydet
kayıt_dosyasi="guncel_veri_seti.csv"
df_guncel.to_csv(kayıt_dosyasi,index=False)
```

```python
#yardımcı fonksiyon: metinleri normalize edelim
#buyuk kucuk harf,bosluk kaldırılması,turkçe kynaklı farkların düzenlenmesi
def metin_normalize_et(series):

  return(
      series
      .str.lower()
      .str.strip()
      .str.replace("ş","s")
      .str.replace("ı","i")
      .str.replace("ç","c")
      .str.replace("ö","o")
  )
```

##Aykırı Değerler(Outliers)
amaç:
1. veri setindeki aykırı değer tespiti ve incelenmesi
2. aykırı değerleri temizleyerek guncel veri seti oluşturulması

Adımlar:
1. veri setini yükle
2. sayısal sutunları belirle
3. sayısal sutunların özet istatistiklerini hesapla
4. IQR ile aykırı değer sınırlarını hesapla
5. Aykırı değer içeren kayıtları tespit etmek
6. aykırı değer sayılarını sutun bazında incelemek
7. aykırı değer içeren kayıtları veri setinden temizle
8. temizlenmiş veri setini kaydet

```python
import pandas as pd
```

```python
#veri setini yükleme
df=pd.read_csv("e_ticaret_veri_seti_tekrarlayan_kayitlar_duzenlendi.csv")
```

```python
#2. sayısal sutun belirlenmesi
sayısal_sutunlar=df.select_dtypes(include=["int64","float64"])
sayısal_sutunlar
```

```python
#3. sayısal sutun özet istatistikleri
df[sayısal_sutunlar.columns].describe()
```

```python
"""birim_siyat
IQR=Q3(%75)-QI(%25)
 =817-198=619
 üst sınır= Q3+K*IQR
 alt sınır=Q1-K*IQR
 k=1 olsun
 üst sınır=1436
 alt sınır=-421
 outlier=üst sınıf ve alt sınıf altında kalanlar
"""
```

```python
# 4. IQR ile aykırı değer sınırlarını hesaplama
aykırı_sınırlar = {}

for sutun in sayısal_sutunlar:
    q1 = df[sutun].quantile(0.25)
    q3 = df[sutun].quantile(0.75)
    iqr = q3 - q1

    üst_sınır = q3 + 1.5 * iqr
    alt_sınır = q1 - 1.5 * iqr

    aykırı_sınırlar[sutun] = {
        "q1": q1,
        "q3": q3,
        "iqr": iqr,
        "üst_sınır": üst_sınır,
        "alt_sınır": alt_sınır
    }
```

```python
alt = aykırı_sınırlar['birim_fiyat']['alt_sınır']
ust = aykırı_sınırlar['birim_fiyat']['üst_sınır']
```

```python
aykırı_sınırlar
```

```python
# 5. Aykırı değer içeren kayıtları tespit et
aykiri_maskesi = pd.Series(False, index=df.index)

for sutun in sayısal_sutunlar:
    alt_sınır = aykırı_sınırlar[sutun]["alt_sınır"]
    üst_sınır = aykırı_sınırlar[sutun]["üst_sınır"]

    sutun_maskesi = (df[sutun] < alt_sınır) | (df[sutun] > üst_sınır)

    aykiri_maskesi = aykiri_maskesi | sutun_maskesi
```

```python
#aykırı değer sayılarını sutun bazında incele
for sutun in sayısal_sutunlar:
    alt_sınır = aykırı_sınırlar[sutun]["alt_sınır"]
    üst_sınır = aykırı_sınırlar[sutun]["üst_sınır"]

    aykiri_sayi = (df[sutun] < alt_sınır) | (df[sutun] > üst_sınır).sum()
    print(f"{sutun}{aykiri_sayi}")

```

```python
#7. aykırı değer içeren kayıtları temizleme
df_temiz=df[~aykiri_maskesi].copy()
df_temiz.shape
```

```python
#8. temizlenmiş veri setini kaydet
df_temiz.to_csv("aykiri_degerler_temizlendi.csv")
```

## Zaman Serisi Analizi
amaç:
1. veri setindeki tarih bilgisini kullanarak zaman bazlı analiz yapmak

adımlar:
1. veri stini yüklemek
2. siparıs tarihini datetime dönüştür
3. tarih sutunundan zaman özellikleri türet
4. aylara göre siparis yogunlugunu incele
5. ayın ilk yarısı ve ikinci yarısındaki harcama farkını incele
6. haftaiçi ve haftasonu siparis davranısı karsılaştırılması
7. iki haftalık dönemlerde ortalama siparıs tutarı incele


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#1. veri seti yüklenmesi
df=pd.read_csv("aykiri_degerler_temizlendi.csv")
df.head()
```

```python
#2. siparıs tarihini datetime dönüştür
df["siparis_tarihi"]=pd.to_datetime(df["siparis_tarihi"])
df["siparis_tarihi"].dtype
```

```python
# 3. Tarih sütunundan zaman özellikleri

df["siparis_yili"] = df["siparis_tarihi"].dt.year
df["siparis_ayi"] = df["siparis_tarihi"].dt.month
df["siparis_gunu"] = df["siparis_tarihi"].dt.day
df["haftanin_gunu"] = df["siparis_tarihi"].dt.day_name()
df["hafta_numarasi"] = df["siparis_tarihi"].dt.isocalendar().week

df.head()
```

```python
#4. aylara göre siparis yogunlugy
aylık_yogunluk=(
    df.groupby("siparis_ayi")
    .agg(
        siparis_sayisı=("siparis_id","count"),
        benzersiz_musteri_sayisi=("musteri_id","nunique"),

    )
)
aylık_yogunluk
```

```python
plt.figure(figsize=(6, 3))
plt.plot(aylık_yogunluk.index,aylık_yogunluk["siparis_sayisı"],marker="o")
plt.title("Aylık Sipariş Sayısı")
plt.xlabel("Aylar")
plt.ylabel("Sipariş Sayısı")
```

```python
#5. ayın ilk ve ikinci yarısındaki harcama farkını inceleme
#ayın ilk yarısı ve ikinci yarısını bulma
df["ay_donemi"]=df["siparis_gunu"].apply(
    lambda x: "ilk_yarı" if x <= 15 else "ikinci_yarı"
)
df.head()
```

```python
ay_donemi_analizi=(
    df.groupby("ay_donemi")
    .agg(
        toplam_tutar=("toplam_tutar","sum"),
        benzersiz_musteri_sayisi=("musteri_id","nunique"),
)
)
ay_donemi_analizi
```

```python
plt.figure(figsize=(6, 3))
plt.bar(ay_donemi_analizi.index,ay_donemi_analizi["toplam_tutar"])
plt.title("Ay Donemi Analizi")
plt.xlabel("Ay Donemi")
plt.ylabel("Toplam Tutar")
```

```python
#6. haftaiçi ve haftasonu siparis davranısını karşılaştırma
df["hafta_tipi"]=df["siparis_tarihi"].dt.dayofweek.apply(
    lambda x: "haftasonu" if x in [5,6] else "haftaiçi"
)
df.head()
```

```python
hafta_tipi_analizi=(
    df.groupby("hafta_tipi")
    .agg(
        ortalama_tutar=("toplam_tutar","mean"),


)
)
hafta_tipi_analizi
```

```python
plt.figure(figsize=(6, 3))
plt.bar(hafta_tipi_analizi.index,hafta_tipi_analizi["ortalama_tutar"])
plt.title("Hafta Tipi Analizi")
plt.xlabel("Hafta Tipi")
plt.ylabel("Ortalama Tutar")
plt.show()
```

```python
# 7. İki haftalık dönemlerde ortalama sipariş tutarı
df["iki_haftalik_donem"] = ((df["hafta_numarasi"] - 1) // 2) + 1
df.head()
```

```python
iki_haftalik_donem_analizi=(
    df.groupby("iki_haftalik_donem")
    .agg(
        ortalama_tutar=("toplam_tutar","mean"),
    )
    .sort_values(by="iki_haftalik_donem")
)
iki_haftalik_donem_analizi
```

```python
plt.figure()
plt.plot(iki_haftalik_donem_analizi.index,iki_haftalik_donem_analizi["ortalama_tutar"])
```

adımlar:

2. hangi ödeme türleri daha yüksek ortalama siparıs tutarı üretiyor ve müşteri puanı açısından performans gösteriyor?

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("aykiri_degerler_temizlendi.csv")
df.head()
```

```python
# 1. Hangi kategori hem yüksek gelir üretir
# hem de müşteri memnuniyeti açısından güçlü görünüyor?

kategori_raporu = df.groupby("kategori").agg(
    toplam_tutar=("toplam_tutar", "sum"),
    ortalama_musteri_puani=("musteri_puani", "mean")
)

kategori_raporu
```

```python
kategori_raporu=kategori_raporu.sort_values(by="toplam_tutar",ascending=False)
kategori_raporu
```

```python
#kategoriye göre toplam gelir
plt.figure(figsize=(6, 3))
plt.bar(kategori_raporu.index,kategori_raporu["toplam_tutar"])
plt.title("Kategoriye Göre Toplam Gelir")
plt.xlabel("Kategori")
plt.ylabel("Toplam Gelir")
plt.show()
```

```python
#kategoriye göre ortalama musteri puanı
plt.figure(figsize=(6, 3))
plt.bar(kategori_raporu.index,kategori_raporu["ortalama_musteri_puani"])
plt.title("Kategoriye Göre Ortalama Müşteri Puanı")
plt.xlabel("Kategori")
plt.ylabel("Ortalama Müşteri Puanı")
plt.show()
```

```python
#2. hangi ödeme türleri daha yüksek ortalama siparıs tutarı üretiyor ve müşteri puanı açısından performans gösteriyor?
odeme_raporu=df.groupby("odeme_turu").agg(
    ortalama_tutar=("toplam_tutar","mean"),
    ortalama_musteri_puani=("musteri_puani","mean")
)
odeme_raporu
```

```python
# ödeme türüne göre ortalama sipariş tutarı
plt.figure(figsize = (8,4))
sns.barplot(data = odeme_raporu, x = "odeme_turu", y = "ortalama_tutar")
plt.title("Ödeme türüne göre ortalama müşteri puanı")
plt.xlabel("Kategori")
plt.ylabel("Müşteri Puanı")
plt.show()
```

```python
# Ödeme türüne göre ortalama müşteri puanı
odeme_raporu = df.groupby("odeme_turu").agg(
    ortalama_musteri_puani=("musteri_puani", "mean")
).reset_index()

plt.figure(figsize=(8, 4))

sns.barplot(
    data=odeme_raporu,
    x="odeme_turu",
    y="ortalama_musteri_puani"
)

plt.title("Ödeme Türüne Göre Ortalama Müşteri Puanı")
plt.xlabel("Ödeme Türü")
plt.ylabel("Ortalama Müşteri Puanı")

plt.show()
```

```python
# 3. Hangi şehir-kategori kombinasyonları en güçlü fırsatı sunuyor?

sehir_kategori_raporu = df.groupby(["sehir", "kategori"]).agg(
    toplam_tutar=("toplam_tutar", "sum")
).sort_values(
    by="toplam_tutar",
    ascending=False
)

sehir_kategori_raporu
```
